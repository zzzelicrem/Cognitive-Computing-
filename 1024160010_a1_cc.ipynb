### Assignment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

n = int(input("enter\n"))
sum = 0

for i in range(1, n + 1):
    if is_prime(i):
        sum += i

print(f"sum: {sum}")
```

### Assignment 6.1: WAP using function that add all odd numbers from 1 to n, n is given by the user.

```python
def sum_odd(n):
    sum = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            sum += i
    return sum

n = int(input("enter\n"))
res = sum_odd(n)
print(f"sum: {res}")
```

### Assignment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def sum_prime(n):
    total = 0
    for i in range(1, n + 1):
        if is_prime(i):
            total += i
    return total

n_sum = int(input("enter\n"))
res = sum_prime(n_sum)
print(f"sum: {res}")
```
