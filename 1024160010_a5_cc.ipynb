## Q2. Basic NumPy Array
(a) Reverse the given array. (b) Find all most-frequent values and their indices in arrays `x` and `y`.

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("Reversed array:", arr[::-1])

def most_frequent_values_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    max_count = counts.max()
    modes = values[counts == max_count]
    return [(value, np.where(array == value)[0]) for value in modes]

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for name, array in [("x", x), ("y", y)]:
    print(f"\nArray {name}: {array}")
    for value, indices in most_frequent_values_and_indices(array):
        print(f"Most frequent value: {value}, frequency: {len(indices)}, indices: {indices.tolist()}")
```

## Q4. NumPy Array Properties and Transpose
Create a 1-D array named `arnav_agarwal` with 25 evenly spaced numbers from 10 to 100. Display its dimensions, shape, number of elements, data type, and memory size. Use `reshape()` to create a column vector and transpose it. Check whether `.T` transposes a 1-D array.

```python
arnav_agarwal = np.linspace(10, 100, 25)

print("Array:", arnav_agarwal)
print("Dimensions:", arnav_agarwal.ndim)
print("Shape:", arnav_agarwal.shape)
print("Total elements:", arnav_agarwal.size)
print("Data type:", arnav_agarwal.dtype)
print("Total bytes:", arnav_agarwal.nbytes)

column_array = arnav_agarwal.reshape(25, 1)
print("\nReshaped column array:")
print(column_array)
print("Transpose using reshape and .T:")
print(column_array.T)

print("\nTranspose of the original 1-D array using .T:")
print(arnav_agarwal.T)
print("Does .T change the 1-D array shape?", arnav_agarwal.T.shape != arnav_agarwal.shape)
```

For a 1-D NumPy array, `.T` does not change its shape. Reshape it into a 2-D array first to obtain a row-to-column or column-to-row transpose.

## Q5. NumPy Array Statistics, Reshape, and Resize
Create a 3 × 4 array named `ucs420_arnav_agarwal`. Calculate its mean, median, maximum, minimum, and unique elements. Reshape it to 4 × 3 and resize it to 2 × 3.

```python
ucs420_arnav_agarwal = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_arnav_agarwal)
print("Mean:", np.mean(ucs420_arnav_agarwal))
print("Median:", np.median(ucs420_arnav_agarwal))
print("Maximum:", np.max(ucs420_arnav_agarwal))
print("Minimum:", np.min(ucs420_arnav_agarwal))
print("Unique elements:", np.unique(ucs420_arnav_agarwal))

reshaped_ucs420_arnav_agarwal = ucs420_arnav_agarwal.reshape(4, 3)
print("\nReshaped array (4 × 3):")
print(reshaped_ucs420_arnav_agarwal)

resized_ucs420_arnav_agarwal = np.resize(ucs420_arnav_agarwal, (2, 3))
print("\nResized array (2 × 3):")
print(resized_ucs420_arnav_agarwal)
```
